## Handling the Missing Value

//Renaming the Column

```python
#.rename() function is used to rename the column
df =df.rename(columns= {'Series_title_1':'SeriesVolume'})
df.head()
```

```python
#in this code the new column is made using df[Value_new] and using data_value column filnna to fill value mean to mean the value and at last typcasting from decimal to integer
df['Value_new']=df['Data_value'].fillna(df['Data_value'].mean()).astype(int)
df.head()
```

```python
#groupby is used to group columns 
grouped_mean =df.groupby('Period')['Data_value'].mean()
print(grouped_mean)
```

```python

grouped_sum = df.groupby(['Series_reference','Subject'])['Data_value'].sum()
print(grouped_sum)
```

```python
# Aggregated multiple function

grouped_agg = df.groupby('Series_reference')['Data_value'].agg(['mean','sum','count'])
grouped_agg
```

## Insight Generation

***The Top Export Growth Products***

```python
# The groupby is used to group, sort_value is used to sort but by default python sort from smaller number to bigger number but by using ascending=False means giving instruction to python to move from higher to lower. 
recent_data = df[df['Period'] == 2024.06]
top_exports = recent_data.groupby('Series_title_2')['Data_value'].mean().sort_values(ascending=False)

print("--- Top Export Categories in 2024 ---")
print(top_exports.head(5))
```

```python
# To find the average export index value across all products
yearly_trend = df.groupby('Period')['Data_value'].mean()

yearly_trend.head(5)
yearly_trend.tail(5)
```

```python
# Find which product groups have the highest variation
volatility = df.groupby('Series_title_2')['Data_value'].std().sort_values(ascending=False)

volatility.head(3)
# .std()= means Standard Deviation
```

```python
#.value_counts() it that how many times each items appears.
status_breakdown = df['STATUS'].value_counts()
status_breakdown
```

```python

#Line charts connect the dots from left to right. This lets your eye instantly spot chronological patterns
# Growth Crashes
trend = df.groupby('Period')['Data_value'].mean()

trend.plot(kind='line', title='Export Trends Over Time')
```

```python
#This kind of layout provide Categorical Rank Comparison
top_products = df.groupby('Series_title_2')['Data_value'].mean().sort_values(ascending=False).head(5)

top_products.plot(kind='bar', title='Top 5 Export Products')
```

```python
# Data Quality Auditing  and proportional analysis
status_counts = df['STATUS'].value_counts()

status_counts.plot(kind='bar', title='Data Status Breakdown')
```
